```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.DataFrame({
    "rollno":[4,3,6,8,3],
    "time":[2,3,4,5,2]
})
print(df)
```

```python
# plt coult take both  array ( anything that is iterable ) in this case this is the series of the pandas
plt.bar(df['rollno'], df['time'])
```

```python
x = [1,4,6,8]
y = [];
for i in x:
    y.append(i**2)
plt.plot(x,y)
plt.xlabel("x axis")
plt.ylabel("y axis")
```

```python
a = plt.figure()
# subplot keeps it default if want to add the axis use the add_axex( values eg left, bottom, width, height)
ax = a.add_subplot()
plt.show() # this will show the ax canvas
```

```python
import seaborn as sns
```

```python
# can load the official datasets ( sample ) provided by the seaborn 
df = sns.load_dataset('tips')
type(df) # we could see that the type is the dataframe of pandas
print(df)
```

```python
# qualitative -> data that is categorical ( eg , male or female , yes or no ) uses like - countplot , pie , bar
# quantitative -> data that is descrete ( continuous vlaue of like anything , age may be that just canges ), uses like histogram , boxplot , line
sns.countplot(data = df, x = 'sex') # this is one of the example
```

```python
df['size'].unique() # so first seeing the size column seems it is quantitative but finding unique we see that it is only up to 1 to 6 so it could be considered into the qualitative 
```

```python
#histplot / distplot same thing
#by default bins is 20
sns.histplot(df['total_bill'], bins = 10)
```

```python
#by default kind is scatter , there are many we could see too
sns.jointplot(x = 'total_bill', y= 'tip', data = df)
```

```python
#sns.pairplot(df)
#hue is for the seperation let say i want to say who tipped most
sns.pairplot(df, hue='sex')
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
```
